# MATERIAL SUPLEMENTARIO: Trazabilidad y Refinamiento del Pipeline Algorítmico (NLP)
**Proyecto:** Brecha editorial en la neuroartropatía de Charcot (2016-2025)

## ACLARACIÓN DE TRANSPARENCIA METODOLÓGICA
En favor del máximo cumplimiento con las directrices de reproducibilidad computacional y ciencia abierta (Open Science), el presente cuaderno desglosa la evolución iterativa del algoritmo de clasificación semántica. El proceso de *Procesamiento de Lenguaje Natural (NLP)* no se ejecutó como una "caja negra" estática, sino mediante un modelo de calibración heurística en dos fases.

El presente documento documenta la transición desde un modelo puramente léxico (Pipeline Inicial) hacia un modelo de estratificación epistemológica basado en reglas de vecindad y contexto causal (Pipeline Consolidado Final).

Aquí el desglose:

Enfoque Descriptivo: El pipeline está diseñado para categorizar la literatura existente. Por ejemplo, el clúster 'B. Mecanicista-biológico' agrupa artículos que investigan rutas metabólicas (RANKL, citoquinas), que son esencialmente estudios sobre la causa biológica del Charcot, pero el código no extrae si esos estudios confirmaron o no la causalidad.

Léxico de Mecanismos: En el Bloque 1.1 y el Motor de Inferencia, verá términos como pathogen, metabolic o pathway. Estos términos identifican artículos que proponen modelos explicativos (causales) del proceso de neuroartropatía.

Limitación: Como este es un pipeline bibliométrico, tratamos los datos como 'incidencias' y 'frecuencias'. No hay un modelo de regresión o inferencia causal ejecutándose en el código.

## FASE 1: PIPELINE CIENTÍFICO CONSOLIDADO INICIAL (N = 364)
**Objetivo:** Exploración basal del rendimiento del motor de clasificación mediante co-ocurrencia léxica simple.

*Nota técnica:* En esta iteración temprana, el universo de $N=364$ artículos primarios fue sometido a un filtrado rígido de palabras clave. El algoritmo identificó artículos como "Sin asignar". La auditoría humana de esta submuestra reveló que la asignación binaria simple era insuficiente para capturar la intencionalidad del autor (ej. el uso de 'neuropathy' como antecedente clínico frente a su uso como foco investigativo). Este hallazgo empírico justificó la necesidad de refinar el código.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
import os

def parse_pubmed_certified(filepath):
    if not os.path.exists(filepath): return []
    articles = []
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read().replace('\x00', '')
        for entry in re.split(r'PMID- ', content)[1:]:
            item = {}
            pmid = re.search(r'^(\d+)', entry)
            item['PMID'] = pmid.group(1) if pmid else None
            ti = re.search(r'TI  - (.*?)(?=\n[A-Z]{2,4}  -)', entry, re.DOTALL)
            item['title'] = ti.group(1).replace('\n      ', ' ').strip() if ti else ''
            jt = re.search(r'JT  - (.*?)(?=\n[A-Z]{2,4}  -)', entry, re.DOTALL)
            item['journal'] = jt.group(1).strip() if jt else 'Unknown'
            ab = re.search(r'AB  - (.*?)(?=\n[A-Z]{2,4}  -)', entry, re.DOTALL)
            item['abstract'] = ab.group(1).replace('\n      ', ' ').strip() if ab else ''
            dp = re.search(r'DP  - (\d{4})', entry)
            item['Año Efectivo'] = int(dp.group(1)) if dp else none
            doi = re.search(r'AID - (.*?) \[doi\]', entry)
            item['DOI'] = doi.group(1) if doi else None
            authors = re.findall(r'FAU - (.*?)(?=\n)', entry)
            item['authors'] = "; ".join(authors) if authors else "Unknown"
            item['Source'] = 'PubMed'
            articles.append(item)
    return articles

df_p = pd.DataFrame(parse_pubmed_certified('/content/pubmed-DiabeticFo-setCOMPLEMENTO.txt') + parse_pubmed_certified('/content/pubmed-DiabeticFoNC-set.txt'))
df_w = pd.read_csv('/content/savedrecsCF.txt', sep='\t').rename(columns={'TI': 'title', 'AB': 'abstract', 'PY': 'Año Efectivo', 'DI': 'DOI', 'PM': 'PMID', 'SO': 'journal', 'AU': 'authors'})
df_w['Source'] = 'Web of Science'
df_s = pd.read_csv('/content/scopus-DiabeticFoNC.csv').rename(columns={'Title': 'title', 'Abstract': 'abstract', 'Year': 'Año Efectivo', 'DOI': 'DOI', 'Source title': 'journal', 'Authors': 'authors'})
df_s['Source'] = 'Scopus'

sjr_neuro = pd.read_csv('/content/scimagojr 2024  Subject Category - Neurology.csv', sep=';')
sjr_clin = pd.read_csv('/content/scimagojr 2024  Subject Category - Neurology (clinical).csv', sep=';')
neuro_list = set(sjr_neuro['Title'].str.upper().unique()) | set(sjr_clin['Title'].str.upper().unique())

def clean_t(t): return re.sub(r'[^a-zA-Z0-9]', '', str(t)).lower()
def clean_j(n): return str(n).split('JID -')[0].strip().upper()

df_raw = pd.concat([df_p, df_w, df_s], ignore_index=True)
priority_map = {'PubMed': 1, 'Scopus': 2, 'Web of Science': 3}
df_raw['priority'] = df_raw['Source'].map(priority_map)
df_raw = df_raw.sort_values(by=['priority', 'PMID'], ascending=True)

df_step1 = df_raw.drop_duplicates(subset=['PMID'], keep='first').drop_duplicates(subset=['DOI'], keep='first')
df_step1['t_norm'] = df_step1['title'].apply(clean_t)
df_certified_f1 = df_step1.drop_duplicates(subset=['t_norm'], keep='first').copy()

noise = r'review|editorial|commentary|reply|letter|systematic|meta-analysis|guideline|consensus|case report|case series'
df_certified_f1 = df_certified_f1[~df_certified_f1['title'].str.contains(noise, case=False, na=False)]
df_certified_f1 = df_certified_f1[(df_certified_f1['abstract'].str.len() > 100) & (df_certified_f1['Año Efectivo'] >= 2016)]
df_certified_f1['journal_clean'] = df_certified_f1['journal'].apply(clean_j)
df_certified_f1['Is_SJR_Neuro'] = df_certified_f1['journal_clean'].isin(neuro_list)

# --- VISUALIZACIÓN ---
DPI = 300
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(4, 1, figsize=(16, 32))

# G1: Evolución Anual
ev_anual = df_certified_f1.groupby('Año Efectivo').size().reindex(range(2016, 2026), fill_value=0)
ev_anual.plot(kind='line', marker='o', color='#34495e', ax=axes[0])
for x, y in ev_anual.items():
    axes[0].annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0,10), ha='center', fontweight='bold')
axes[0].set_title(f'G1: Evolución Anual (N={len(df_certified_f1)})', fontweight='bold', fontsize=16)
axes[0].set_xlabel('Año Efectivo', fontweight='bold')
axes[0].set_ylabel('Frecuencia', fontweight='bold')

# G2: Top 10 Autores con etiquetas por pila
all_authors = []
for _, row in df_certified_f1.iterrows():
    if pd.notnull(row['authors']):
        for a in str(row['authors']).split(';'): all_authors.append({'Author': a.strip(), 'Año': row['Año Efectivo']})
df_auth = pd.DataFrame(all_authors)
top_10_a = df_auth['Author'].value_counts().head(10).index
auth_pivot = df_auth[df_auth['Author'].isin(top_10_a)].groupby(['Año', 'Author']).size().unstack(fill_value=0).reindex(range(2016, 2026), fill_value=0)
auth_pivot.plot(kind='bar', stacked=True, ax=axes[1], colormap='tab20')

for n, col in enumerate(auth_pivot.columns):
    bottoms = auth_pivot.iloc[:, :n].sum(axis=1)
    for i, val in enumerate(auth_pivot[col]):
        if val > 0:
            axes[1].text(i, bottoms.iloc[i] + val/2, f'{int(val)}', ha='center', va='center', fontweight='bold', color='black', fontsize=9)

axes[1].set_title(f'G2: Top 10 Autores (Frecuencia Acumulada)', fontweight='bold', fontsize=16)
axes[1].set_xlabel('Año', fontweight='bold')
axes[1].set_ylabel('Artículos', fontweight='bold')
axes[1].legend(bbox_to_anchor=(1.05, 1), loc='upper left', frameon=True)

# G3: Top 10 Revistas
top_10_j = df_certified_f1['journal_clean'].value_counts().head(10).index
jrnl_pivot = df_certified_f1[df_certified_f1['journal_clean'].isin(top_10_j)].groupby(['Año Efectivo', 'journal_clean']).size().unstack(fill_value=0).reindex(range(2016, 2026), fill_value=0)
jrnl_pivot.plot(kind='line', marker='s', ax=axes[2], linewidth=2)
axes[2].set_title('G3: Top 10 Revistas - Producción Anual', fontweight='bold', fontsize=16)
axes[2].set_xlabel('Año Efectivo', fontweight='bold')
axes[2].set_ylabel('N', fontweight='bold')
axes[2].legend(bbox_to_anchor=(1.05, 1), loc='upper left', frameon=True)

# G4: Disparidad Editorial
df_neuro_only = df_certified_f1[df_certified_f1['Is_SJR_Neuro'] == True]
top_neuro_series = df_neuro_only['journal_clean'].value_counts().head(1)
top_neuro_name = top_neuro_series.index[0] if not top_neuro_series.empty else "NINGUNA"
general_counts = df_certified_f1['journal_clean'].value_counts().head(10)
final_plot_data = pd.concat([general_counts, top_neuro_series]) if top_neuro_name != "NINGUNA" and top_neuro_name not in general_counts.index else general_counts
colors_g4 = ['#e74c3c' if j == top_neuro_name else '#bdc3c7' for j in final_plot_data.index]
final_plot_data.plot(kind='bar', ax=axes[3], color=colors_g4, logy=True)
for i, v in enumerate(final_plot_data):
    axes[3].text(i, v, f' {int(v)}', va='bottom', ha='center', fontweight='bold', color='black')
axes[3].set_title('G4: Disparidad Editorial (Escala Log)', fontweight='bold', fontsize=16)
axes[3].set_xlabel('Revista', fontweight='bold')
axes[3].set_ylabel('Frecuencia (Log)', fontweight='bold')

plt.tight_layout()
# Guardar en alta calidad
plt.savefig('Graficos_Fase1_Consolidados.jpg', dpi=DPI, bbox_inches='tight')
plt.show()
```

## FASE 2: PIPELINE CIENTÍFICO CONSOLIDADO FINAL (N = 464)
**Objetivo:** Despliegue del motor de inferencia avanzado con reglas de dependencia sintáctica y filtro de ruido documental.

*Nota técnica:* Tras la re-extracción y deduplicación exhaustiva que garantizó la mayor sensibilidad de las bases de datos ($N=464$), se implementó el algoritmo refinado. Las modificaciones incluyeron:
1. **Regla de Vecindad:** Restricción de términos neurológicos a contextos diagnósticos/funcionales explícitos, pero no de forma estricta (basada en distancia de palabras), pero sí de forma semántica.

En el código actual, los términos de la categoría C (re_C) ya están filtrados para ser mayoritariamente diagnósticos o funcionales (ej. nerve conduction, electromyography, thermography, sudomotor). Esto evita que una mención genérica a 'nervio' clasifique el artículo.

Sin embargo, el motor actual busca estos términos en cualquier parte del abstract.

2. **Estratificación Epistemológica:** Integración de la categoría 'Híbrido' (H) para identificar estudios integradores, y redefinición de 'Sin asignar' (S) como estudios fenotípico-descriptivos.

```python
import pandas as pd
import re
import os

# --- CONFIGURACIÓN MAESTRA MeSH (Descriptor: Arthropathy, Neurogenic) ---
# Incluimos el descriptor oficial y sus Entry Terms (Charcot, etc.)
pattern_charcot_core = r'charcot|neuroarthropathy|neuropathic arthropathy|neurogenic arthropath|joint[\s,]+charcot'

def parse_pubmed_certified(filepath):
    if not os.path.exists(filepath): return []
    articles = []
    with open(filepath, 'r', encoding='utf-8') as f:
        for entry in re.split(r'PMID- ', f.read())[1:]:
            item = {}
            pmid = re.search(r'^(\d+)', entry)
            item['PMID'] = pmid.group(1) if pmid else None
            ti = re.search(r'TI  - (.*?)(?=\n[A-Z]{2,4}  -)', entry, re.DOTALL)
            item['title'] = ti.group(1).replace('\n      ', ' ').strip() if ti else ''
            ab = re.search(r'AB  - (.*?)(?=\n[A-Z]{2,4}  -)', entry, re.DOTALL)
            item['abstract'] = ab.group(1).replace('\n      ', ' ').strip() if ab else ''
            dp = re.search(r'DP  - (\d{4})', entry)
            item['Año Efectivo'] = int(dp.group(1)) if dp else None
            doi = re.search(r'AID - (.*?) [doi]', entry)
            item['DOI'] = doi.group(1) if doi else None
            jt = re.search(r'JT  - (.*?)(?=\n[A-Z]{2,4}  -)', entry, re.DOTALL)
            item['journal'] = jt.group(1).strip() if jt else 'Unknown'
            item['Source'] = 'PubMed'
            articles.append(item)
    return articles

try:
    df_p = pd.DataFrame(parse_pubmed_certified('/content/pubmed-DiabeticFo-setCOMPLEMENTO.txt') + parse_pubmed_certified('/content/pubmed-DiabeticFoNC-set.txt'))
    df_w = pd.read_csv('/content/savedrecsCF.txt', sep='\t').rename(columns={'TI': 'title', 'AB': 'abstract', 'PY': 'Año Efectivo', 'DI': 'DOI', 'SO': 'journal'})
    df_s = pd.read_csv('/content/scopus-DiabeticFoNC.csv').rename(columns={'Title': 'title', 'Abstract': 'abstract', 'Year': 'Año Efectivo', 'DOI': 'DOI', 'Source title': 'journal'})

    df_all = pd.concat([df_p, df_w.assign(Source='Web of Science'), df_s.assign(Source='Scopus')], ignore_index=True, sort=False)
    df_all['abstract'] = df_all['abstract'].fillna('')

    def strict_validation_v2(row):
        title_abs = (str(row['title']) + ' ' + str(row['abstract'])).lower()
        # EXCLUSIÓN RIGUROSA (CMT, Spine, etc.)
        exclusion_terms = ['hand', 'wrist', 'spine', 'spinal', 'charcot-marie-tooth', 'cmt', 'syphilis', 'spondyloarthritis', 'leprosy', 'alcoholism']
        if any(term in title_abs for term in exclusion_terms): return False

        # FILTRO DE ESTUDIOS NO PRIMARIOS / RUIDO
        noise = r'review|meta-analysis|systematic review|bibliometric|guideline|consensus|editorial|commentary|reply|letter'
        if re.search(noise, str(row['title']).lower()): return False

        # TRIPLE VALIDACIÓN (DESCRIPTOR MeSH: Arthropathy, Neurogenic + DIABETES + ANATOMY)
        has_core = bool(re.search(pattern_charcot_core, title_abs))
        has_context = bool(re.search(r'diabet|dm1|dm2|diabetes mellitus', title_abs))
        has_anatomy = bool(re.search(r'foot|ankle|pedis|tarsal|lower limb|toe', title_abs))
        return has_core and has_context and has_anatomy

    df_all['passed'] = df_all.apply(strict_validation_v2, axis=1)
    df_certified = df_all[(df_all['passed'] == True) & (df_all['Año Efectivo'] >= 2016) & (df_all['abstract'].str.len() > 150)].copy()
    df_certified['t_norm'] = df_certified['title'].apply(lambda t: "".join(filter(str.isalnum, str(t))).lower())
    df_certified = df_certified.sort_values('Source', ascending=False).drop_duplicates(subset=['t_norm'], keep='first').copy()

    print(f"Pipeline MeSH Certificado (Arthropathy, Neurogenic): N={len(df_certified)} artículos primarios.")
except Exception as e:
    print(f"Error: {e}")

    # 1. Auditoría de Atrición (Pérdida de registros)
n_total_inicial = len(df_all)

# Calculamos pasos intermedios para la auditoría
# 1. Filtro de Ruido (Tipo de documento/Longitud)
df_noise_passed = df_all[df_all['abstract'].str.len() > 150].copy()
n_post_ruido = len(df_noise_passed)

# 2. Filtro de Semántica (Basado en la columna 'passed' del pipeline)
n_post_semantica = df_all['passed'].sum()

# 3. Certificado Final (Post-Deduplicación)
n_final_certificado = len(df_certified)

print("--- REPORTE DE INTEGRIDAD DEL PIPELINE ---")
print(f"Total registros brutos (PubMed+WoS+Scopus): {n_total_inicial}")
print(f"Excluidos por Ruido (Abstracts cortos/Tipo): {n_total_inicial - n_post_ruido}")
print(f"Excluidos por Semántica (No Pie/No Diabet/CMT/Hand): {n_post_ruido - n_post_semantica}")
print(f"Excluidos por Deduplicación Cross-Database: {n_post_semantica - n_final_certificado}")
print(f"\nUniverso Certificado Final (N): {n_final_certificado}")

# Muestra de excluidos con DOI, PMID y Título (sin resumen)
df_excluded = df_all[df_all['passed'] == False].sample(min(5, len(df_all[df_all['passed'] == False])))
print("\nMuestra de artículos EXCLUIDOS para verificación (DOI, PMID, Título):")
display(df_excluded[['DOI', 'PMID', 'title']].head())
```

```python
import pandas as pd
import re

# --- MOTOR DE INFERENCIA ---
def classify_v4_clean(row):
    title = str(row['title']).lower()
    abstract = str(row['abstract']).lower()
    text = title + ' ' + abstract

    # 1. DEFINICIÓN DE PATRONES TEMÁTICOS (A/B/C)
    re_A = r'surgery|amputation|debridement|reconstruction|arthrodesis|off-loading|palliative|osteotomy|fixation|ulcer|wound'
    re_B = r'bone metabolism|osteoclast|osteoblast|rankl|osteoprotegerin|metabolic|bone resorption|inflammation|tnf-alpha|interleukin'
    re_C = r'neuropathy|nerve conduction|electromyography|thermography|autonomic|sensory|denervation|small fiber|sudomotor|vibration'

    # CONTEO DE HITS
    hits_A = len(re.findall(re_A, text))
    hits_B = len(re.findall(re_B, text))
    hits_C = len(re.findall(re_C, text))

    # LÓGICA DE CATEGORIZACIÓN FINAL
    # Si hay solapamiento significativo de temas, es Híbrido
    if (hits_B >= 1 and hits_C >= 1) or (hits_B >= 1 and hits_A >= 1) or (hits_C >= 1 and hits_A >= 1):
        return "Híbrido"

    if hits_A == 0 and hits_B == 0 and hits_C == 0:
        return "Sin asignar"

    # Clasificación por predominancia
    if hits_B > hits_A and hits_B >= hits_C: return "B. Mecanicista-Biológico"
    if hits_C > hits_A and hits_C > hits_B: return "C. Neurológico-Fisiológico"
    if hits_A >= hits_B and hits_A >= hits_C: return "A. Clínico-Correctivo"

    return "Sin asignar"

# Aplicar la nueva clasificación limpia
df_certified['Categoría_V2'] = df_certified.apply(classify_v4_clean, axis=1)

print("Distribución de Categorías (Versión Limpia):")
display(df_certified['Categoría_V2'].value_counts())
```

```python
import pandas as pd
import re

# --- CONFIGURACIÓN UNIFICADA DE PATRONES (FULL MeSH LIST) ---
pattern_charcot_core = r'charcot|neuroarthropathy|neuropathic arthropathy|neurogenic arthropath|joint[\s,]+charcot'
re_A = r'\bsurg|\bamput|\bulcer|\barthrod|\breconstruct|wound|off[\s-]?load'
re_B = r'\bpathogen|\bsorbitol|rankl|osteoprotegerin|metabolic|cytokine'
re_C = r'\bthermog|electromyogr|nerve[\s-]conduction|vibration|sensory|autonomic|denervation|fiber'

def count_semantic_category_hits(df_subset, re_A_pat, re_B_pat, re_C_pat):
    if df_subset.empty: return "0/0/0"
    text_combined = (df_subset['title'].astype(str) + ' ' + df_subset['abstract'].astype(str)).str.lower()
    hits_A_count = text_combined.apply(lambda t: bool(re.search(re_A_pat, t))).sum()
    hits_B_count = text_combined.apply(lambda t: bool(re.search(re_B_pat, t))).sum()
    hits_C_count = text_combined.apply(lambda t: bool(re.search(re_C_pat, t))).sum()
    return f"{hits_A_count}/{hits_B_count}/{hits_C_count}"

def get_category_percentages_robust(df_subset):
    if df_subset.empty: return {}
    total_articles = len(df_subset)
    # Normalizamos los nombres de las categorías encontradas para sumar correctamente
    counts = df_subset['Categoría_V2'].value_counts()
    mapping = {
        'A. Clínico-correctivo': 0.0,
        'B. Mecanicista-biológico': 0.0,
        'C. Neurológico-neurofisiológico': 0.0,
        'Híbrido': 0.0,
        'Sin asignar': 0.0
    }

    for real_name, count in counts.items():
        low_name = str(real_name).lower()
        if 'a.' in low_name: mapping['A. Clínico-correctivo'] += count
        elif 'b.' in low_name: mapping['B. Mecanicista-biológico'] += count
        elif 'c.' in low_name: mapping['C. Neurológico-neurofisiológico'] += count
        elif 'híbrido' in low_name: mapping['Híbrido'] += count
        else: mapping['Sin asignar'] += count

    return {k: round((v / total_articles * 100), 1) for k, v in mapping.items()}

report_data = []
sources = ['PubMed', 'Web of Science', 'Scopus', 'TOTAL CONSOLIDADO']
categories_for_table = ['A. Clínico-correctivo', 'B. Mecanicista-biológico', 'C. Neurológico-neurofisiológico', 'Híbrido', 'Sin asignar']

for source in sources:
    df_sub = df_certified.copy() if source == 'TOTAL CONSOLIDADO' else df_certified[df_certified['Source'] == source].copy()
    cat_percentages = get_category_percentages_robust(df_sub)
    text_full = (df_sub['title'].astype(str) + ' ' + df_sub['abstract'].astype(str)).str.lower()
    menciones_charcot_n = text_full.str.contains(pattern_charcot_core, case=False, na=False).sum()
    hits_semantic_abc = count_semantic_category_hits(df_sub, re_A, re_B, re_C)

    row = {'Fuente': source, 'N Total (Certificado)': len(df_sub)}
    for cat in categories_for_table:
        row[f'% {cat[:12]}...'] = cat_percentages.get(cat, 0.0)
    row.update({'Menciones Charcot (N)': menciones_charcot_n, 'Hits (A/B/C)': hits_semantic_abc})
    report_data.append(row)

super_tabla_1 = pd.DataFrame(report_data)
print("### TABLA 1: DISTRIBUCIÓN POR FUENTE Y MÉTRICAS (ROBUST MAPPING) ###")
fmt_func = lambda x: f"{int(x)}" if isinstance(x, float) and x % 1 == 0 else f"{x:.1f}"
display(super_tabla_1.style.format(fmt_func, subset=[c for c in super_tabla_1.columns if '%' in c]).set_properties(**{'text-align': 'left'}).hide(axis='index'))

```

```python
import pandas as pd
import re
from collections import Counter

# --- TABLA 2: PROPORCIÓN DE MENCIONES DE CHARCOT POR CATEGORÍA ---
# Reutilizamos el patrón MeSH y los términos semánticos definidos arriba

def get_top_terms(df_sub, pattern, top_n=3):
    all_text = (df_sub['title'].astype(str) + ' ' + df_sub['abstract'].astype(str)).str.lower()
    words = []
    for text in all_text:
        found = re.findall(pattern, text)
        words.extend([f.strip() for f in found if len(f) > 3])
    counts = Counter(words)
    return ", ".join([f"{word} ({count})" for word, count in counts.most_common(top_n)])

report_data = []
# Definimos las categorías exactas según el motor V2
categories_certified = ['A. Clínico-Correctivo', 'B. Mecanicista-Biológico', 'C. Neurológico-Fisiológico', 'Híbrido', 'Sin asignar']

patt_top = "|".join([re_A, re_B, re_C])

for c_key in categories_certified:
    sub = df_certified[df_certified['Categoría_V2'] == c_key].copy()
    n_cat = len(sub)
    if n_cat == 0: continue

    text_full_cat = (sub['title'].astype(str) + ' ' + sub['abstract'].astype(str)).str.lower()
    charcot_hits = text_full_cat.str.contains(pattern_charcot_core, case=False, na=False).sum()

    def get_source_metrics(db_name, df_subset):
        db_sub = df_subset[df_subset['Source'] == db_name]
        if len(db_sub) == 0: return "-"
        text_db = (db_sub['title'].astype(str) + ' ' + db_sub['abstract'].astype(str)).str.lower()
        hit = text_db.str.contains(pattern_charcot_core, case=False, na=False).sum()
        val = round((hit/len(db_sub))*100, 1)
        return f"{int(val) if val % 1 == 0 else val}%"

    report_data.append({
        'Categoría': c_key,
        'N': n_cat,
        '% Global': (n_cat / len(df_certified) * 100),
        'Mención Core (n)': charcot_hits,
        '% Core/Cat': (charcot_hits / n_cat * 100) if n_cat > 0 else 0,
        'Top Términos (Freq)': get_top_terms(sub, patt_top) if c_key != 'Sin asignar' else "-",
        'PubMed (%Ch)': get_source_metrics('PubMed', sub),
        'WoS (%Ch)': get_source_metrics('Web of Science', sub),
        'Scopus (%Ch)': get_source_metrics('Scopus', sub)
    })

super_tabla_2 = pd.DataFrame(report_data)
print(f"TABLA 2: PROPORCIÓN DE MENCIONES DE CHARCOT POR CATEGORÍA (N={len(df_certified)})")
print("Nota: '% Core' incluye Charcot y sus sinónimos MeSH.")

fmt_func = lambda x: f"{int(x)}" if isinstance(x, float) and x % 1 == 0 else f"{x:.1f}"
display(super_tabla_2.style.format(fmt_func, subset=['% Global', '% Core/Cat']).set_properties(**{'text-align': 'left'}).hide(axis='index'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from fuzzywuzzy import process
import re
from matplotlib.ticker import FixedLocator

# 1. NORMALIZACIÓN Y CARGA DE IMPACTO
def clean_name(name):
    if not name: return ""
    name = re.sub(r'\(.*?\)', '', str(name))
    return " ".join(name.lower().split())

def load_sjr_data():
    try:
        sjr_n = pd.read_csv('/content/scimagojr 2024  Subject Category - Neurology.csv', sep=';')
        sjr_c = pd.read_csv('/content/scimagojr 2024  Subject Category - Neurology (clinical).csv', sep=';')
        sjr_n['SJR_Cat_Ref'] = 'Neurology'
        sjr_c['SJR_Cat_Ref'] = 'Neurology (clinical)'
        combined = pd.concat([sjr_n, sjr_c]).drop_duplicates(subset=['Title'], keep='first')
        combined['Title_Clean'] = combined['Title'].apply(clean_name)
        return combined
    except: return pd.DataFrame()

df_sjr_ref = load_sjr_data()
df_viz = df_certified[(df_certified['Año Efectivo'] >= 2016) & (df_certified['Año Efectivo'] <= 2025)].copy()

col_q = [c for c in df_sjr_ref.columns if 'Quartile' in c or 'Best Q' in c]
col_q = col_q[0] if col_q else None

# 2. MOTOR DE CRUCE REFINADO
if not df_sjr_ref.empty and col_q:
    sjr_lookup = dict(zip(df_sjr_ref['Title_Clean'], df_sjr_ref.index))
    sjr_titles_clean = df_sjr_ref['Title_Clean'].tolist()

    def match_impact_refined(journal_name):
        c_name = clean_name(journal_name)
        if not c_name or c_name == 'unknown': return pd.Series(['Sin Q', 'Fuera de Neuro'])
        if c_name in sjr_lookup:
            row = df_sjr_ref.loc[sjr_lookup[c_name]]
            return pd.Series([str(row[col_q]).strip(), str(row['SJR_Cat_Ref'])])
        match, score = process.extractOne(c_name, sjr_titles_clean)
        if score > 85:
            row = df_sjr_ref[df_sjr_ref['Title_Clean'] == match].iloc[0]
            q_val = row[col_q] if pd.notnull(row[col_q]) else 'Sin Q'
            return pd.Series([str(q_val).strip(), str(row['SJR_Cat_Ref'])])
        return pd.Series(['Sin Q', 'Fuera de Neuro'])

    df_viz[['SJR_Q', 'SJR_Cat']] = df_viz['journal'].apply(match_impact_refined)
else:
    df_viz['SJR_Q'], df_viz['SJR_Cat'] = 'Sin Q', 'Fuera de Neuro'

df_viz['Categoría_V2'] = df_viz['Categoría_V2'].astype(str)
df_viz['SJR_Cat'] = df_viz['SJR_Cat'].apply(lambda x: 'Neurology' if 'clinical' not in str(x).lower() and 'fuera' not in str(x).lower() else ('Neurology (clinical)' if 'clinical' in str(x).lower() else 'Fuera de Neuro'))
df_viz['SJR_Q'] = df_viz['SJR_Q'].replace('nan', 'Sin Q')

# 3. LÓGICA TEMÁTICA D3
pattern_charcot = r'charcot|neuroarthropathy|neuropathic arthropathy|neurogenic arthropath|joint[\s,]+charcot'
df_viz['Is_Charcot_MeSH'] = (df_viz['title'].str.lower() + ' ' + df_viz['abstract'].str.lower()).str.contains(pattern_charcot, na=False, regex=True)

term_map = {
    'Surgery/Fixation': r'surg|fixation|reconstruct|amputat',
    'Ulcer/Wound': r'ulcer|wound|infect',
    'Neuropathy/Nerve': r'neuropath|nerve|sensory'
}

for label, pat in term_map.items():
    df_viz[label] = (df_viz['title'].str.lower() + ' ' + df_viz['abstract'].str.lower()).str.contains(pat, na=False, regex=True)

# CONFIGURACIÓN VISUAL CONSOLIDADA (300 DPI)
plt.style.use('seaborn-v0_8-whitegrid')
cat_col = 'Categoría_V2'
DPI = 300
order_cats_desc = df_viz[cat_col].value_counts().index
order_cats_asc = df_viz[cat_col].value_counts().index[::-1]

fig, axes = plt.subplots(4, 1, figsize=(15, 36))

# --- D1: Mapa de calor ---
ct = pd.crosstab(df_viz[cat_col], df_viz['Año Efectivo']).reindex(order_cats_desc)
sns.heatmap(ct, annot=True, fmt='d', cmap='YlGnBu', ax=axes[0])
axes[0].set_title('D1: Distribución Anual por Clúster', fontweight='bold', fontsize=16)
axes[0].set_xlabel('Año Efectivo', fontweight='bold')
axes[0].set_ylabel('Categoría', fontweight='bold')
axes[0].set_yticklabels([t.get_text().replace('_', ' ') for t in axes[0].get_yticklabels()])

# --- D2: Calidad Editorial ---
sns.countplot(data=df_viz, x=cat_col, order=order_cats_asc, hue='SJR_Q', hue_order=['Q1', 'Q2', 'Q3', 'Q4', 'Sin Q'], palette='RdYlGn_r', ax=axes[1])
axes[1].set_title('D2: Calidad Editorial (SJR Q) por Categoría', fontweight='bold', fontsize=16)
axes[1].set_xlabel('Categoría', fontweight='bold')
axes[1].set_ylabel('Cantidad de Artículos', fontweight='bold')
axes[1].xaxis.set_major_locator(FixedLocator(axes[1].get_xticks()))
axes[1].set_xticklabels([t.get_text().replace('_', ' ') for t in axes[1].get_xticklabels()], rotation=15)
axes[1].legend(title='Cuartil SJR', frameon=True, facecolor='white', framealpha=1, edgecolor='black')
for p in axes[1].patches:
    if p.get_height() > 0:
        axes[1].annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width()/2., p.get_height()), ha='center', va='center', xytext=(0, 7), textcoords='offset points', fontsize=10, fontweight='bold')

# --- D3: Evolución Incidencia (CON VALORES) ---
years = np.arange(2016, 2026)
counts_charcot = df_viz[df_viz['Is_Charcot_MeSH'] == True].groupby('Año Efectivo').size().reindex(years, fill_value=0)
axes[2].plot(years, counts_charcot.values, marker='o', label='Menciones Charcot (Core)', linewidth=5, color='#e74c3c', zorder=5)
# Agregar valores para Charcot Core
for x, y in zip(years, counts_charcot.values):
    axes[2].annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0,10), ha='center', fontweight='bold', color='#e74c3c')

colors_trends = ['#2ecc71', '#f1c40f', '#3498db']
for (label, _), color in zip(term_map.items(), colors_trends):
    counts = df_viz[df_viz[label] == True].groupby('Año Efectivo').size().reindex(years, fill_value=0)
    line = axes[2].plot(years, counts.values, marker='s', label=label, linewidth=2, linestyle='--', color=color, alpha=0.8)
    # Agregar valores para cada tendencia
    for x, y in zip(years, counts.values):
        axes[2].annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0,-15), ha='center', fontsize=9, fontweight='bold', color=color)

axes[2].set_title('D3: Evolución por Incidencia Temática (No Excluyente)', fontweight='bold', fontsize=16)
axes[2].set_xlabel('Año de Publicación', fontweight='bold')
axes[2].set_ylabel('N (Incidencia)', fontweight='bold')
axes[2].set_xticks(years)
axes[2].legend(frameon=True, facecolor='white', framealpha=1, edgecolor='black', fontsize=11, loc='upper left')

# --- D4: Disparidad ---
sns.countplot(data=df_viz, x=cat_col, order=order_cats_asc, hue='SJR_Cat', palette='viridis', hue_order=['Neurology', 'Neurology (clinical)', 'Fuera de Neuro'], ax=axes[3])
axes[3].set_title('D4: Disparidad (Clúster vs Revista)', fontweight='bold', fontsize=16)
axes[3].set_xlabel('Categoría', fontweight='bold')
axes[3].set_ylabel('Cantidad de Artículos', fontweight='bold')
axes[3].xaxis.set_major_locator(FixedLocator(axes[3].get_xticks()))
axes[3].set_xticklabels([t.get_text().replace('_', ' ') for t in axes[3].get_xticklabels()], rotation=15)
axes[3].legend(title='Categoría SJR', frameon=True, facecolor='white', framealpha=1, edgecolor='black')
for p in axes[3].patches:
    if p.get_height() > 0:
        axes[3].annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width()/2., p.get_height()), ha='center', va='center', xytext=(0, 7), textcoords='offset points', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('Graficos_Fase2_Consolidados.jpg', dpi=DPI, bbox_inches='tight')
plt.show()
```

### 📖 Guía de Interpretación: Gráfico D3 (Evolución de Incidencia Temática)

Este gráfico es el pilar de la prueba del **"Editorial Gap"**. Aquí se detalla cómo leer sus componentes y la lógica algorítmica detrás de cada serie:

#### 1. Lógica de Incidencia No Excluyente (Solapamiento)
*   **¿Qué mide?**: A diferencia de los gráficos categóricos (donde 1 artículo = 1 categoría), el D3 mide la **densidad terminológica**.
*   **El Fenómeno del Solapamiento**: Si un artículo científico se titula *"Surgical reconstruction of the neuropathic foot"*, el algoritmo detecta hits tanto para `Cirugía` como para `Neuropatía`. En este gráfico, ese artículo **suma +1 en ambas líneas**.
*   **Interpretación del Gap**: Si la línea de `Cirugía/Fijación` está muy por encima de la de `Neuropatía/Nervio`, significa que aunque los autores reconozcan el origen neurológico (mencionan el término), el foco discursivo y el volumen de producción están capturados por la intervención mecánica/quirúrgica.

#### 2. La Línea Maestra (Core MeSH)
*   **Color Rojo (Menciones Core Charcot)**: Representa los artículos que cumplen estrictamente con los descriptores oficiales (MeSH) de la enfermedad. Actúa como el "techo" de relevancia; cualquier serie que se acerque a esta línea domina el campo.

#### 3. Diccionario de 'Top Words' (Patrones de Búsqueda)
El gráfico lee el texto del título y el resumen buscando clústeres de palabras clave:

*   **🔵 Cirugía/Fijación**: Busca la frecuencia de intervención activa (*surg, fixation, reconstruct, amputat*). Representa el enfoque correctivo.
*   **🟡 Úlcera/Infección**: Busca la complicación paliativa (*ulcer, wound, infect*). Es el marcador de manejo clínico tradicional.
*   **🟢 Neuropatía/Nervio**: Busca el origen neurofisiológico (*neuropath, nerve, sensory*).

#### 4. Cómo leer las tendencias
*   **Divergencia**: Si la distancia entre la línea azul (Cirugía) y la verde (Neurología) aumenta con los años, la brecha editorial se está ensanchando.
*   **Convergencia**: Si las líneas se acercan, indica una tendencia hacia la investigación multidisciplinar o 'Híbrida'.
*   **Volumen Total**: Es normal que la suma de las incidencias sea mayor al N=464 del estudio, precisamente porque este gráfico visibiliza la **coexistencia** de conceptos en un mismo texto.

Lógica de Incidencia No Excluyente:

Independencia: Si un artículo describe una técnica quirúrgica para tratar una úlcera neuropática, ese registro sumará +1 en las tres series (Surgery, Ulcer y Neuropathy).
Realidad Clínica: Esto evita forzar al artículo a elegir una sola 'etiqueta', permitiendo que el gráfico D3 muestre cómo co-evolucionan los temas. Por eso verás que, si sumas las líneas, el total puede ser mayor a 464; es el reflejo del solapamiento temático.
Fidelidad: De este modo, no 'robamos' relevancia a una categoría para dársela a otra; mostramos la presencia real de cada concepto en el corpus bibliográfico.

### 🛡️ CONTROL DE CALIDAD POST-AUDITORÍA (Estándares de Ciencia Abierta)
Basado en el dictamen del revisor, procedemos a verificar que los filtros aplicados (especialmente el de longitud de abstract) no introdujeron sesgos sistemáticos entre las fuentes (PubMed, WoS, Scopus).

```python
import pandas as pd
import re

if 'df_all' in locals():
    # 1. Re-definición de Patrones Maestros
    pattern_charcot_core = r'charcot|neuroarthropathy|neuropathic arthropathy|neurogenic arthropath|joint[\s,]+charcot'

    def strict_validation_v2(row):
        title_abs = (str(row['title']) + ' ' + str(row['abstract'])).lower()
        # EXCLUSIÓN RIGUROSA COMPLETA
        exclusion_terms = ['hand', 'wrist', 'spine', 'spinal', 'charcot-marie-tooth', 'cmt', 'syphilis', 'spondyloarthritis', 'leprosy', 'alcoholism']
        if any(term in title_abs for term in exclusion_terms): return False

        # FILTRO DE ESTUDIOS NO PRIMARIOS
        noise = r'review|meta-analysis|systematic review|bibliometric|guideline|consensus|editorial|commentary|reply|letter'
        if re.search(noise, str(row['title']).lower()): return False

        # TRIPLE VALIDACIÓN MeSH/ANATOMÍA/CONTEXTO
        has_core = bool(re.search(pattern_charcot_core, title_abs))
        has_context = bool(re.search(r'diabet|dm1|dm2|diabetes mellitus', title_abs))
        has_anatomy = bool(re.search(r'foot|ankle|pedis|tarsal|lower limb|toe', title_abs))
        return has_core and has_context and has_anatomy

    # 2. Aplicar validación y filtros de calidad
    df_all['passed'] = df_all.apply(strict_validation_v2, axis=1)
    df_passed = df_all[(df_all['passed'] == True) & (df_all['Año Efectivo'] >= 2016) & (df_all['abstract'].str.len() > 150)].copy()

    # Normalización idéntica a la original para deduplicación
    df_passed['t_norm'] = df_passed['title'].apply(lambda t: "".join(filter(str.isalnum, str(t))).lower())

    # 3. Análisis de Exclusividad
    counts_per_title = df_passed.groupby('t_norm')['Source'].nunique()
    titles_exclusive = counts_per_title[counts_per_title == 1].index
    df_exclusive = df_passed[df_passed['t_norm'].isin(titles_exclusive)]
    exclusividad = df_exclusive['Source'].value_counts().to_frame(name='Artículos Únicos')

    # 4. Aplicar Prioridad Estricta (PubMed > Scopus > WoS)
    priority_map = {'PubMed': 1, 'Scopus': 2, 'Web of Science': 3}
    df_passed['priority'] = df_passed['Source'].map(priority_map)
    df_certified = df_passed.sort_values('priority', ascending=True).drop_duplicates(subset=['t_norm'], keep='first').copy()

    # 5. Tabla de Atrición
    final_counts = df_certified['Source'].value_counts()
    atricion = df_all.groupby('Source').agg(
        Brutos=('title', 'count'),
        Cortos_Excluidos=('abstract', lambda x: (x.str.len() <= 150).sum())
    )
    atricion['Final_Certificado'] = atricion.index.map(lambda x: final_counts.get(x, 0))
    atricion['%_Retención'] = (atricion['Final_Certificado'] / atricion['Brutos'] * 100).round(2)

    # Mostrar resultados finales
    print(f"--- 1. TABLA DE ATRICIÓN N={len(df_certified)} ---")
    display(atricion)

    print("\n--- 2. TABLA DE EXCLUSIVIDAD ---")
    display(exclusividad)

    n_overlap = len(df_passed) - len(df_certified)
    print(f"\nResumen: {len(df_passed)} pasaron filtros. {n_overlap} solapados. {len(df_certified)} final.")

    # 6. Generación de archivo final Excel si df_certified existe
    if 'df_certified' in locals() and not df_certified.empty:
        cols_final = [c for c in ['DOI', 'PMID', 'Año Efectivo', 'Source', 'Categoría_V2', 'title', 'journal'] if c in df_certified.columns]
        df_certified[cols_final].to_excel('Charcot_Database_Final_CERTIFIED_V2.xlsx', index=False)
        print("\n✅ Archivo 'Charcot_Database_Final_CERTIFIED_V2.xlsx' generado con éxito.")
    else:
        print("\n⚠️ Advertencia: 'df_certified' está vacío o no se generó correctamente.")

else:
    print("❌ ERROR: 'df_all' no encontrado. Por favor ejecuta el Pipeline de la Fase 2 para cargar los datos brutos.")
```

### 📌 Hoja de Ruta del Algoritmo – Fase 2 (Motor de Inferencia V2)

#### 1. Principio de Triple Validación MeSH
Para que un artículo ingrese al Universo Certificado (N=464), el algoritmo aplicó un filtro booleano `AND` estricto en el abstract:
*   **Patología:** Presencia de descriptores core (`Charcot`, `Neuroarthropathy`).
*   **Contexto:** Presencia de términos metabólicos de base (`Diabetes`, `DM1`, `DM2`).
*   **Anatomía:** Localización específica (`Foot`, `Ankle`, `Pedis`).

#### 2. Jerarquía de Clasificación (Lógica de Predominancia)
El algoritmo cuenta 'hits' (menciones) de términos clave y asigna la categoría según el peso relativo:

*   **🔷 A — Clínico-Correctivo:**
    *   *Criterio:* Predominio de términos quirúrgicos o de manejo de úlceras.
    *   *Keywords:* `surgery`, `reconstruction`, `amputation`, `off-loading`, `ulcer`.

*   **🔷 B — Mecanicista-Biológico:**
    *   *Criterio:* Enfoque en la fisiopatología molecular y remodelado óseo.
    *   *Keywords:* `RANKL`, `osteoclast`, `metabolic`, `cytokine`, `inflammation`.

*   **🔷 C — Neurológico-Fisiológico:**
    *   *Criterio:* Enfoque en la función nerviosa y diagnóstico neurofisiológico.
    *   *Keywords:* `nerve conduction`, `sudomotor`, `small fiber`, `electromyography`.

*   **🔷 H — Híbrido:**
    *   *Regla de Solapamiento:* Si el artículo presenta $\ge 1$ hit en dos o más categorías distintas, el algoritmo lo etiqueta automáticamente como **Híbrido**. Esto captura la naturaleza multidisciplinar del estudio.

*   **🔷 S — Sin asignar (Descriptivo):**
    *   *Criterio:* Artículos que pasan el filtro MeSH pero cuya densidad de términos A, B o C es cero.
    *   *Naturaleza:* Estudios epidemiológicos puros, reportes de casos sin técnica quirúrgica nueva o validación de escalas diagnósticas generales.

#### 3. Reglas de Exclusión Automática
1.  **Ruido Documental:** Eliminación de `Review`, `Editorial` y `Letter` mediante Regex en el título.
2.  **Longitud Crítica:** Exclusión de abstracts < 150 caracteres para evitar falta de poder estadístico en la clasificación.
3.  **Fuera de Foco:** Filtro `NOT` para `Charcot-Marie-Tooth (CMT)`, `Spine` y `Hand`.

```python
import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score, confusion_matrix
from scipy import stats

# --- CARGA Y PREPARACIÓN DE DATOS (Arreglo del NameError) ---
try:
    # Cargamos el archivo de validación directamente para asegurar su existencia
    df_validacion = pd.read_excel('/content/Validación de muestra.xlsx', header=0, usecols="E:G").dropna(how='all')
    df_validacion.columns = ['Eval_E', 'Eval_T', 'Motor_V2_Excel']

    # Limpieza y estandarización
    df_validacion = df_validacion.map(lambda x: str(x).strip() if pd.notnull(x) else '0')

    # Creamos df_analisis (solo casos temáticos sin exclusiones '0')
    df_analisis = df_validacion[(df_validacion['Eval_E'] != '0') & (df_validacion['Eval_T'] != '0')].copy()

    print(f"✅ Datos cargados: N_Bruto={len(df_validacion)}, N_Temático={len(df_analisis)}")
except Exception as e:
    print(f"❌ Error crítico al cargar los datos: {e}")

# --- FUNCIONES ESTADÍSTICAS ---
def kappa_stats(y_true, y_pred, labels):
    k = cohen_kappa_score(y_true, y_pred, labels=labels)
    n = len(y_true)
    se = np.sqrt((k * (1 - k)) / n) if (k > 0 and n > 0) else 1e-9
    ci_low = k - (1.96 * se)
    ci_high = k + (1.96 * se)
    z = k / se if se > 0 else 0
    p_val = 2 * (1 - stats.norm.cdf(abs(z)))

    if k < 0: interp = "Sin acuerdo"
    elif k <= 0.2: interp = "Insignificante (Slight)"
    elif k <= 0.4: interp = "Aceptable (Fair)"
    elif k <= 0.6: interp = "Moderado (Moderate)"
    elif k <= 0.8: interp = "Sustancial (Substantial)"
    else: interp = "Casi Perfecto (Almost Perfect)"

    return {"Kappa": round(k, 3), "SE": round(se, 3), "IC_95": f"[{round(ci_low,3)}, {round(ci_high,3)}]", "p-val": f"{p_val:.4f}", "Interp": interp}

def print_matrix_discrepancia(y_alg, y_hum, name):
    cats = ['A', 'B', 'C', 'H', 'S']
    cm = confusion_matrix(y_alg, y_hum, labels=cats)
    df_cm = pd.DataFrame(cm, index=[f'Máquina: {c}' for c in cats], columns=[f'Humano: {c}' for c in cats])
    df_cm['Total Máquina'] = df_cm.sum(axis=1)
    total_humano = df_cm.sum(axis=0).to_frame().T
    total_humano.index = ['Total Humano']
    final_cm = pd.concat([df_cm, total_humano])
    print(f"\n--- MATRIZ DE DISCREPANCIA: Algoritmo vs {name} ---")
    display(final_cm)

# --- ESCENARIOS ---
if 'df_validacion' in locals():
    labels_all = ['A', 'B', 'C', 'H', 'S', '0']
    print("### ESCENARIO A: VALIDACIÓN BRUTA (INCLUYENDO '0') ###")
    results_a = {
        "Eval_E vs Eval_T": kappa_stats(df_validacion['Eval_E'], df_validacion['Eval_T'], labels_all),
        "Eval_E vs Algoritmo": kappa_stats(df_validacion['Eval_E'], df_validacion['Motor_V2_Excel'], labels_all),
        "Eval_T vs Algoritmo": kappa_stats(df_validacion['Eval_T'], df_validacion['Motor_V2_Excel'], labels_all)
    }
    display(pd.DataFrame(results_a).T)

    print(f"\n### ESCENARIO B: CONCORDANCIA TEMÁTICA (N={len(df_analisis)}) ###")
    labels_clean = ['A', 'B', 'C', 'H', 'S']
    results_b = {
        "Humano E vs Humano T": kappa_stats(df_analisis['Eval_E'], df_analisis['Eval_T'], labels_clean),
        "Algoritmo vs Humano E": kappa_stats(df_analisis['Eval_E'], df_analisis['Motor_V2_Excel'], labels_clean),
        "Algoritmo vs Humano T": kappa_stats(df_analisis['Eval_T'], df_analisis['Motor_V2_Excel'], labels_clean)
    }
    display(pd.DataFrame(results_b).T)

    print_matrix_discrepancia(df_analisis['Motor_V2_Excel'], df_analisis['Eval_E'], 'Evaluador E')
    print_matrix_discrepancia(df_analisis['Motor_V2_Excel'], df_analisis['Eval_T'], 'Evaluador T')
```
